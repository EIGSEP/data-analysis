```python
from datetime import datetime
import glob
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

import eigsep_observing as eo

%matplotlib widget
```

```python
DIR = "/media/christian/Samsung_T5/eigsep_data/deployment5/"
```

```python
lidar = []
times = []
for p in Path(DIR).glob("corr_20260718*"):
    d, h, m = eo.io.read_hdf5(p)
    rt = h["run_tag"]
    if not rt == "motor_scan":
        continue
    times.extend(h["times"])
    lidar_dat = m.get("lidar")
    if lidar_dat is None:
        print('missing')
        continue
    for ld in lidar_dat:
        if ld is None:
            dist = 0
        else:
            dist = ld.get("distance_m", 0)
        lidar.append(dist)
```

```python
plt.figure()
plt.plot(times)
plt.show()

print(np.argmax(np.diff(times)))
```

```python
plt.figure()
plt.plot(times[:720]-times[0], lidar[:720])
plt.grid()
plt.xlabel(f"Seconds since {datetime.fromtimestamp(times[0])}")
plt.show()

plt.figure()
plt.plot(times[720:]-times[720], lidar[720:])
plt.grid()
plt.xlabel(f"Seconds since {datetime.fromtimestamp(times[720])}")
plt.show()
```

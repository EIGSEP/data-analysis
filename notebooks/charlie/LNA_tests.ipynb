```python
import matplotlib.pyplot as plt
import numpy as np
import csv
import os
from cmt_vna import calkit
%matplotlib notebook
```

```python
file = './LNA_TESTS/LNA01.csv'
def get_data(file):
    with open(file, 'r') as file:
        reader = csv.reader(file, delimiter=',')
        freqs = []
        data = []
        dataset = {}
        for row in reader:
            if row[0] == '! DATA Freq':
                datatype =row[1]

            elif row[0] == 'END':
                dataset[datatype] = np.array(data)
                data = []

            else:
                try:
                    freq = float(row[0])
                    dat = float(row[1])
                    freqs.append(freq)
                    data.append(dat)
                except ValueError:
                    continue
    return np.array(freqs), dataset
```

```python
freqs, dataset = get_data(file)
```

```python
sparams = ['S11', 'S12', 'S21', 'S22']
for i in sorted(os.listdir('./LNA_TESTS/')):
    if '.csv' in i:
        fig, axes = plt.subplots(nrows=2, ncols=2, sharex = True, sharey=True)
        file = os.path.join('./LNA_TESTS', i)
        freqs, dataset = get_data(file)
        axes[0][0].plot(freqs[:801]/1e6, dataset['S11'])
        axes[0][0].text(0,20,'S11',bbox=dict(facecolor='white', alpha=0.5))
        axes[0][1].plot(freqs[:801]/1e6, dataset['S12'])
        axes[0][1].text(0,20,'S12',bbox=dict(facecolor='white', alpha=0.5))
        axes[1][0].plot(freqs[:801]/1e6, dataset['S21'])
        axes[1][0].text(450,5,'S21',bbox=dict(facecolor='white', alpha=0.5))
        axes[1][1].plot(freqs[:801]/1e6, dataset['S22'])
        axes[1][1].text(0,20,'S22',bbox=dict(facecolor='white', alpha=0.5))
        plt.suptitle(i[:-4])
        plt.tight_layout()
        plt.savefig(f'./LNA_TESTS/{i[:-4]}.jpg', dpi=200)
        plt.show()
```

```python
plt.figure(figsize=(9,4))
for i in sorted(os.listdir('./LNA_TESTS/')):
    if '.csv' in i:
        file = os.path.join('./LNA_TESTS', i)
        freqs, dataset = get_data(file)
        plt.plot(freqs[:801]/1e6, dataset['S21'], label=i[:-4])
plt.legend(ncols=2)
plt.grid()
plt.title('Gain')
plt.ylim(0)
plt.xlabel('Freqs (MHz)')
plt.ylabel('gain (dB)')
plt.xlim(-10, 300)
plt.tight_layout()
plt.savefig('LNA_tests/LNA_test_gains.jpg', dpi = 200)
plt.show()
```
